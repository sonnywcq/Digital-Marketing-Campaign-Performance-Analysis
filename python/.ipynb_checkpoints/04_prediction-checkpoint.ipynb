# 04. Purchase Propensity Prediction (Optional)

Predict whether a user will make a purchase after the model's feature window. A time-based split is used to reduce leakage: features are built before a cutoff date, and the target is purchase activity after the cutoff.

```python

from pathlib import Path
import warnings
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    roc_auc_score, average_precision_score,
    precision_score, recall_score, f1_score,
    confusion_matrix, ConfusionMatrixDisplay,
    classification_report
)

warnings.filterwarnings("ignore")

DATA_DIR = Path("../data/processed")
OUTPUT_DIR = Path("../data/analysis")
MODEL_DIR = Path("../models")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
MODEL_DIR.mkdir(parents=True, exist_ok=True)

users = pd.read_csv(DATA_DIR / "users_clean.csv").drop_duplicates("user_id")
events = pd.read_csv(DATA_DIR / "ad_events_clean.csv")

events["event_datetime"] = pd.to_datetime(events["event_datetime"], errors="coerce")
events["event_date"] = pd.to_datetime(events["event_date"], errors="coerce")

print(events["event_datetime"].min(), "to", events["event_datetime"].max())
```

```python

# ========== 1. Define Feature Window and Target Window ==========
min_date = events["event_datetime"].min()
max_date = events["event_datetime"].max()

cutoff = min_date + (max_date - min_date) * 0.70
print("Feature cutoff:", cutoff)

feature_events = events[events["event_datetime"] < cutoff].copy()
future_events = events[events["event_datetime"] >= cutoff].copy()

# Target: user purchases at least once in the future window
future_purchase = (
    future_events.groupby("user_id")["event_type"]
    .apply(lambda x: (x == "Purchase").any())
    .astype(int)
    .rename("future_purchase")
)

# Include every known user
model_df = users[["user_id"]].drop_duplicates().merge(
    future_purchase,
    on="user_id",
    how="left"
)
model_df["future_purchase"] = model_df["future_purchase"].fillna(0).astype(int)

print(model_df["future_purchase"].value_counts(normalize=True))
```

```python

# ========== 2. Build Pre-Cutoff Behavioral Features ==========
feature_events["is_impression"] = (feature_events["event_type"] == "Impression").astype(int)
feature_events["is_click"] = (feature_events["event_type"] == "Click").astype(int)
feature_events["is_engagement"] = feature_events["event_type"].isin(
    ["Like", "Share", "Comment"]
).astype(int)
feature_events["is_purchase"] = (feature_events["event_type"] == "Purchase").astype(int)

features = (
    feature_events.groupby("user_id")
    .agg(
        impressions=("is_impression", "sum"),
        clicks=("is_click", "sum"),
        engagements=("is_engagement", "sum"),
        historical_purchases=("is_purchase", "sum"),
        active_days=("event_date", "nunique"),
        total_events=("event_id", "count"),
        last_event=("event_datetime", "max")
    )
    .reset_index()
)

features["recency_days"] = (cutoff - features["last_event"]).dt.total_seconds() / 86400
features = features.drop(columns=["last_event"])

features["CTR"] = np.where(
    features["impressions"] > 0,
    features["clicks"] / features["impressions"],
    0
)

features["historical_CVR"] = np.where(
    features["clicks"] > 0,
    features["historical_purchases"] / features["clicks"],
    0
)

model_df = model_df.merge(features, on="user_id", how="left")

# Numeric demographics where available
for col in ["age"]:
    if col in users.columns:
        model_df = model_df.merge(
            users[["user_id", col]].drop_duplicates("user_id"),
            on="user_id",
            how="left",
            suffixes=("", "_user")
        )

feature_cols = [
    "impressions",
    "clicks",
    "engagements",
    "historical_purchases",
    "active_days",
    "total_events",
    "recency_days",
    "CTR",
    "historical_CVR"
]

if "age" in model_df.columns:
    feature_cols.append("age")

X = model_df[feature_cols].copy()
y = model_df["future_purchase"].copy()

display(model_df[["user_id", "future_purchase"] + feature_cols].head())
print("Features:", feature_cols)
```

```python

# ========== 3. Train / Test Split ==========
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Train:", X_train.shape, "Test:", X_test.shape)
print("Positive rate - train:", y_train.mean())
print("Positive rate - test:", y_test.mean())
```

```python

# ========== 4. Models ==========
logistic_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ))
])

rf_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", RandomForestClassifier(
        n_estimators=300,
        max_depth=8,
        min_samples_leaf=10,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    ))
])

models = {
    "Logistic Regression": logistic_model,
    "Random Forest": rf_model
}

results = []
fitted_models = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    proba = model.predict_proba(X_test)[:, 1]
    pred = (proba >= 0.50).astype(int)

    results.append({
        "model": name,
        "ROC_AUC": roc_auc_score(y_test, proba),
        "PR_AUC": average_precision_score(y_test, proba),
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred, zero_division=0),
        "F1": f1_score(y_test, pred, zero_division=0)
    })
    fitted_models[name] = model

model_results = pd.DataFrame(results).sort_values("ROC_AUC", ascending=False)
display(model_results)
model_results.to_csv(OUTPUT_DIR / "prediction_model_results.csv", index=False)
```

```python

# ========== 5. Confusion Matrix for Best Model ==========
best_name = model_results.iloc[0]["model"]
best_model = fitted_models[best_name]

best_pred = best_model.predict(X_test)
cm = confusion_matrix(y_test, best_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Purchase", "Purchase"]
)
disp.plot()
plt.title(f"Confusion Matrix - {best_name}")
plt.tight_layout()
plt.show()

print(classification_report(
    y_test,
    best_pred,
    target_names=["No Purchase", "Purchase"],
    zero_division=0
))
```

```python

# ========== 6. Feature Importance ==========
if best_name == "Random Forest":
    rf = best_model.named_steps["model"]
    importance = pd.DataFrame({
        "feature": feature_cols,
        "importance": rf.feature_importances_
    }).sort_values("importance", ascending=False)
else:
    lr = best_model.named_steps["model"]
    importance = pd.DataFrame({
        "feature": feature_cols,
        "importance": np.abs(lr.coef_[0])
    }).sort_values("importance", ascending=False)

display(importance)

plt.figure(figsize=(9, 5))
top_imp = importance.head(10).sort_values("importance")
plt.barh(top_imp["feature"], top_imp["importance"])
plt.title(f"Top Features - {best_name}")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

importance.to_csv(OUTPUT_DIR / "prediction_feature_importance.csv", index=False)
```

```python

# ========== 7. Save Best Model ==========
joblib.dump(
    {
        "model": best_model,
        "features": feature_cols,
        "cutoff": cutoff
    },
    MODEL_DIR / "purchase_propensity_model.joblib"
)

print("Saved:", MODEL_DIR / "purchase_propensity_model.joblib")
```

## Interpretation Note

This model estimates purchase propensity from pre-cutoff behavior. It should be presented as an analytical exercise rather than a production targeting model. Model performance can change with the sample, feature window, class imbalance, and data quality.
